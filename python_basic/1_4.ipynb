```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
year = np.arange(2009, 2019, 1)
year
#从2009开始，在小于2019的范围内，每次增加1
```

```python
year = np.linspace(2009, 2018, 10)
year
# 从2009开始到2018结束，等间隔的分割成10个数
# 这次的year会把之前定义的year覆盖掉
```

```python
temperature = np.array([15, 12, 13, 19, 14, 14, 15, 16, 10, 17])
temperature
```

```python
plt.plot(year, temperature) # 绘制折线图 year是横坐标，temperature是纵坐标
plt.show() # 输出图形
```

```python
plt.scatter(year, temperature) # 绘制散点图
plt.show()
```

```python
plt.bar(year, temperature) # 画柱形图
plt.show()
```

```python
plt.plot(year, temperature, c='k') #将折线变成黑色，c指定颜色，以下是对折线图的修改
plt.show()
```

```python
plt.plot(year, temperature, linewidth=10) # linewidth自定义折线宽度
plt.show()
```

```python
plt.scatter(year, temperature, marker='+') # 将点变成加号，以下是对散点图的修改
plt.show()
```

```python
plt.scatter(year, temperature, marker='^', c='r', alpha = 0.5, s = 500)
plt.show()
# 多个限定条件可以一起用
```

```python
# 设定图表的坐标轴
plt.plot(year, temperature)
plt.xlim([1990, 2025]) # x轴的范围设定为1990-2025
plt.ylim([5, 25]) # y轴的范围设定为5-25
plt.show()
```

```python
# 修改坐标轴的显示（两个点之间相隔几）
plt.plot(year, temperature)
# 设定间隔
tick_x = np.arange(2009, 2019, 1) #横坐标两个点间隔1
tick_y = np.arange(8, 21, 3) # 纵坐标两个点间隔3
plt.xticks(tick_x)
plt.yticks(tick_y)

plt.show()
```

```python
# 设置坐标轴的名字
plt.plot(year, temperature)
plt.xlabel('Year') #横坐标名字设置成year
plt.ylabel('Temperature (°C)') # 纵坐标名字设置成temperature
plt.show()
```

```python
plt.plot(year, temperature, c = 'b', linewidth = 3)
#设定坐标范围
plt.xlim([2009, 2019])
plt.ylim([9, 20])

# 设置点间距和字体font
tick_x = np.arange(2009, 2020, 3)
tick_y = np.arange(10, 20, 3)
plt.xticks(tick_x, fontname = 'Helvetica', fontsize = 15)
plt.yticks(tick_y, fontname='Helvetica', fontsize = 15)

# 设置坐标轴名字和字体
plt.xlabel('Year', fontname = 'Helvetica', fontsize = 18)
plt.ylabel('Temperature(°C)', fontname = 'Helvetica', fontsize = 18)
# 上下左右留白
plt.subplots_adjust(left=0.5, right=0.85, bottom=0.15, top=0.85)
# 将图表保存成图像的形式
plt.savefig('temperature.png', format='png', dpi=300)

plt.show()
```
