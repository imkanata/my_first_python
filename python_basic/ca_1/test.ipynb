```python
# D问题
import datetime

# 始まりの日と終わるの日を設定する
start_date = datetime.date(1979, 1, 1)
end_date = datetime.date(2021, 12, 31)

total = 0
current_date = start_date

while current_date <= end_date:
    date_str = current_date.strftime('%Y%m%d')
    cnt = date_str.count('9') # 9の数

    discount = cnt * 30
    total += (640 - discount)
    
    st = False
    if current_date.year == 2021 and current_date.month == 2 and current_date.month == 2:
        st = True
    elif current_date.year in [1980, 1984] and current_date.month == 2 and current_date.month == 4:
        st = True
    elif current_date.year not in [1980, 1984, 2021] and current_date.month == 2 and current_date.month == 3:
        st = True

    if st :
        total += (540 - discount)

    current_date += datetime.timedelta(days=1)

print(total)
    
```
