```python
import numpy as np
A = np.array([[1, 2, 3], [4, 2, 5], [7, 4, 9]]) #定义矩阵A
A #输出A
```

```python
A.shape #输出A的大小
```

```python
# X = np.array([[1, 2], [2, 3, 5],[6]]) 不均匀会报错
# X 输出x
```

```python
A[1, 2] #下标是（1,2），同C++下标从0开始
```

```python
A[0, :] #输出横坐标为0的行
```

```python
A[: ,1] #输出纵坐标为1的列
```

```python
A[:,0:2] #从纵坐标为0的列开始，往后输出两列
```

```python
A[:,1:3]
```

```python
A[0:2,1] #行从0开始往下找两行，列固定为下标1
#虽然在意义上输出的是列，但是在计算机上用行显示
```

```python
B=A.T #B是A的转置矩阵
B #输出B
```

```python
print(A)
B
print(B[1, 2])
#如果不加print的话，默认只输出最后一行的变量（只写变量名）
# 想强制输出什么的话，就用print
```

```python
A+B #矩阵A+矩阵B的和
```

```python
D = np.linalg.inv(A) #输出A的逆矩阵
D
```

```python
A@D #矩阵A和D的积
```

```python
np.dot(A,D) #矩阵A和D的积
```

```python
A*D # A和D的对应位置相乘
# 比如A的（1,1）位置的数和D（1,1）位置的数相乘就是答案的（1,1）位置的数
```

```python
np.linalg.eig(A) #A的固有值和对应的固有向量
```
