```python
# A问题
import numpy as np
A = np.array([[3, 2, 4], [4, 2, 5], [7, 2, 9]])
B = A.T # B是A的转置
C_1 = np.dot(A, B)
C_2 = np.dot(B, A)
print(C_1)
print(C_2)
```

```python
# B问题
import matplotlib.pyplot as plt
time = np.array([2, 23, 17, 11, 9, 14, 21])
score = np.array([10, 17, 12, 4, 13, 18, 11])

plt.xlim([1, 30])
plt.ylim([2, 20])

tick_x = np.arange(0,32,2)
tick_y = np.arange(1,20,3)
plt.xticks(tick_x, fontname='Arial', fontsize=15)
plt.yticks(tick_y, fontname='Arial',fontsize=15)

plt.xlabel('Time(min)',fontname='Arial',fontsize=16)
plt.ylabel('Score',fontname='Arial',fontsize=16)

plt.subplots_adjust(left=0.15, right=0.85, bottom=0.15, top=0.85)

plt.savefig('testB.png',format='png',dpi=300)

plt.scatter(time, score)

plt.show()
```
