```python
# 2.2列表定义和下标
temperature  = [18, 20, 22, 25, 21] #创建了一个列表存储五天的温度
#下标从0开始 temperature[0] = 18
print(temperature[0])
print(temperature[-1]) #支持负数索引，从列表末尾开始
```

```python
# 2.3列表切片
# list[start:end] 包含start不包含end
temperature = [18, 20, 22, 25, 21]
print(temperature[1:4]) # 下标从1开始到4之前结束
print(temperature[:3]) # 从列表最左端开始到end-1结束
print(temperature[2:]) # 从2开始到列表最右端结束
```

```python
# range()和for循环
for i in range(5):
    print(i)
# range(stop) 生成从0到stop-1
# range(start, stop) 生成start到stop-1
# range(start, stop, step) 从start开始，每次增加step，最大不超过stop-1

temperature = [18, 20, 22, 25, 21]
for t in temperature:
    print(t)
```

```python
# 函数
def celsius_to_fahrenheit(celsius):
    fahrenheit = celsius * 9 / 5 + 32
    return fahrenheit
# def 定义函数
# celsius_to_fahrenheit 函数名
# celsius 传的参数
# return 返回值

result = celsius_to_fahrenheit(20) # 调用函数
print(result)
```
