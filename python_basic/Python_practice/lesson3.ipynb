```python
temperature = [18, 22, 15, 27]
# 字典 key:value
# key用于标识数据，value是实际存储的数据
temperature = {
    "Monday": 18,
    "Tuesday": 22,
    "Wednesday": 15,
    "Thursday": 27
}

# 访问字典里面的数据 字典名称["key"]
print(temperature["Monday"])
# 也可以使用get去输出，找不到对应的key的时候返回None
print(temperature.get("Friday"))

# 添加数据
temperature["Friday"] = 30
# 如果key已经存在，就会修改对应的value
temperature["Monday"] = 20

# 删除数据
del temperature["Friday"]
# 也可以使用pop，会返回被删除的value
t = temperature.pop("Monday")
print(t)
print(temperature)

# 使用for去遍历字典
weather = {
    "Monday": 18,
    "Tuesday": 22,
    "Wednesday": 15,
    "Thursday": 27
}
# 只遍历key,以下两种方法等价
for day in weather:
    print(day)
for day in weather.keys():
    print(day)

# 只遍历value
for temp in weather.values():
    print(temp)

# 同时遍历key和value
# 每次循环去除一组key和value 分别放入day temp
for day, temp in weather.items():
    print(day, temp)
```
