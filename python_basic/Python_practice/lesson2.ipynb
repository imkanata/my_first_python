```python
temperature = [15, 18, 21, 24, 27]

# 修改列表中的元素
temperature[1] = 19
print(temperature)

# 在列表末尾添加元素
temperature.append(26)
print(temperature)

# 在指定位置插入元素 List.insert(index, value)
temperature.insert(2, 22) # 在索引2的位置插入22，原本的元素会往后移动
print(temperature)

# 根据元素的值删除
# List.remove(value)找到第一个等于value的值并删除
# 列表中不存在要删的值的话会报错ValueError
temperature.remove(21) 
print(temperature)

# 根据索引删除并返回元素
removed1 = temperature.pop(2)
removed2 = temperature.pop() # 不指定下标的话默认删除最后一个元素
print(removed1)
print(removed2)

# 列表长度
print(len(temperature))

# 列表排序
temperature.sort() # 从小到大排序
print(temperature)
temperature.sort(reverse=True) # 从大到小排序
print(temperature)

# 反转列表
numbers = [10, 9, 20, 40]
numbers.reverse() # 只进行反转，不进行排序
print(numbers)
```
